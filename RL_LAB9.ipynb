```python

# Assignment 1: Bellman Expectation Equations
# Stochastic policy and deterministic environment

gamma = 0.9

# Bellman equations:
# V(S0) = 0.9*(-1 + gamma*V(S1)) + 0.1*(-1)
# V(S1) = 0.9*(100) + 0.1*(-1 + gamma*V(S0))

# Simplified equations:
# V0 = -1 + 0.81*V1
# V1 = 89.9 + 0.09*V0

# Solve algebraically
V0 = (-1 + 0.81 * 89.9) / (1 - 0.81 * 0.09)
V1 = 89.9 + 0.09 * V0

# Calculate action values
Q_S0_Down = -1 + gamma * V1
Q_S0_Right = -1
Q_S1_Right = 100
Q_S1_Up = -1 + gamma * V0

print("Assignment 1 Results")
print(f"V(S0)       = {V0:.2f}")
print(f"V(S1)       = {V1:.2f}")
print(f"Q(S0, Down) = {Q_S0_Down:.2f}")
print(f"Q(S0, Right)= {Q_S0_Right:.2f}")
print(f"Q(S1, Right)= {Q_S1_Right:.2f}")
print(f"Q(S1, Up)   = {Q_S1_Up:.2f}")

# Verify state values
print("\nVerification:")
print(f"V(S0) = {0.9*Q_S0_Down + 0.1*Q_S0_Right:.2f}")
print(f"V(S1) = {0.9*Q_S1_Right + 0.1*Q_S1_Up:.2f}")
```

```python

# Assignment 2: Iterative Policy Evaluation

gamma = 0.9
theta = 0.000001

V0 = 0.0
V1 = 0.0
iterations = 0

# Iterative policy evaluation
while True:
    new_V0 = 0.1 * (-1 + gamma * 0) + \
             0.9 * (-1 + gamma * V1)

    new_V1 = 0.9 * (100 + gamma * 0) + \
             0.1 * (-1 + gamma * V0)

    delta = max(abs(new_V0 - V0), abs(new_V1 - V1))

    V0, V1 = new_V0, new_V1
    iterations += 1

    if delta < theta:
        break

# Calculate action values
Q_S0_Down = 0.1 * (-1) + 0.9 * (-1 + gamma * V1)
Q_S0_Right = -1
Q_S1_Right = 0.9 * 100 + 0.1 * (-1 + gamma * V0)
Q_S1_Up = -1 + gamma * V0

# Display results
print("Assignment 2 Results")
print("Iterations:", iterations)
print(f"V(S0)       = {V0:.6f}")
print(f"V(S1)       = {V1:.6f}")
print(f"Q(S0, Down) = {Q_S0_Down:.6f}")
print(f"Q(S0, Right)= {Q_S0_Right:.6f}")
print(f"Q(S1, Right)= {Q_S1_Right:.6f}")
print(f"Q(S1, Up)   = {Q_S1_Up:.6f}")

# Policy consistency checks
print("\nPolicy Consistency Checks")
print("V(S0) = Q(S0, Down):",
      abs(V0 - Q_S0_Down) < 1e-6)
print("V(S1) = Q(S1, Right):",
      abs(V1 - Q_S1_Right) < 1e-6)
```
