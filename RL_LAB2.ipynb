```python
import random

# Machines
machines = ['A', 'B', 'C']

# Initialize
N = [0, 0, 0]
R = [0, 0, 0]
Q = [0, 0, 0]

epsilon = 0.2
total_reward = 0

# 10 plays
for play in range(10):

    r = random.random()

    # Exploration
    if r < epsilon:
        index = random.randint(0, 2)
        print("Exploration")

    # Exploitation
    else:
        index = Q.index(max(Q))
        print("Exploitation")

    print("Play:", play + 1)
    print("Selected Machine:", machines[index])

    # Input reward
    reward = int(input("Enter reward: "))

    # Update N
    N[index] += 1

    # Update R
    R[index] += reward

    # Update Q
    Q[index] = Q[index] + (reward - Q[index]) / N[index]

    # Total reward
    total_reward += reward

# Final output
print("\nFinal Results")
print("Machine\tN\tR\tQ")

for i in range(3):
    print(machines[i], "\t", N[i], "\t", R[i], "\t", round(Q[i], 2))

print("\nTotal Reward =", total_reward)
```
